```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
ecoli = pd.read_csv("../data/amr_ecoli.csv")
kleb = pd.read_csv("../data/amr_klebsiella.csv")
```

```python
ecoli.head()
```

```python
kleb.head()
```

```python
ecoli.columns
```

```python
def clean_amr(df):
    df = df[df["Resistant Phenotype"].isin(["Susceptible", "Resistant"])]
    df["label"] = df["Resistant Phenotype"].map({
        "Susceptible": 0,
        "Resistant": 1
    })
    return df

ecoli_clean = clean_amr(ecoli)
kleb_clean = clean_amr(kleb)
```

```python
ecoli_clean["label"].value_counts(normalize=True)
kleb_clean["label"].value_counts(normalize=True)
```

```python
def resistance_rate(df):
    return df.groupby("Antibiotic")["label"].mean().sort_values(ascending=False)

ecoli_rates = resistance_rate(ecoli_clean)
kleb_rates = resistance_rate(kleb_clean)
```

```python
amr_compare = pd.DataFrame({
    "E_coli": ecoli_rates,
    "K_pneumoniae": kleb_rates
}).dropna()
```

```python
plt.figure()
plt.imshow(amr_compare.T, aspect="auto")
plt.xticks(range(len(amr_compare)), amr_compare.index, rotation=90)
plt.yticks([0,1], ["E. coli", "K. pneumoniae"])
plt.colorbar(label="Resistance Frequency")
plt.title("Comparative Antimicrobial Resistance Patterns")
plt.tight_layout()
plt.show()
```
