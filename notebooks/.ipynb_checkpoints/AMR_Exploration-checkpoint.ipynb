## Introduction

Antimicrobial resistance (AMR) is one of the most pressing global health
challenges, particularly in low- and middle-income countries where
infectious diseases remain a major cause of morbidity and mortality.
Among Gram-negative bacteria, *Escherichia coli* and *Klebsiella pneumoniae*
are of particular concern due to their frequent involvement in community-
and hospital-acquired infections and their increasing resistance to
multiple classes of antibiotics.

In routine clinical microbiology laboratories, antimicrobial susceptibility
testing (AST) is performed daily to guide treatment decisions. However,
beyond individual patient reports, aggregated susceptibility data can
provide valuable insights into resistance trends, organism-specific
patterns, and emerging threats. Systematic analysis of such data is essential
for antimicrobial stewardship, surveillance, and research.

This project aims to perform a comparative analysis of antimicrobial
resistance patterns in *E. coli* and *K. pneumoniae* using curated phenotypic
susceptibility data. In addition to classical exploratory data analysis,
basic machine learning techniques are used to assist in identifying
antibiotics and resistance patterns that most strongly differentiate the
two organisms.

### Objectives of the Study

1. To explore and compare resistance frequencies of *E. coli* and
   *Klebsiella pneumoniae* across commonly tested antibiotics.
2. To identify antibiotics that show distinct resistance behavior between
   the two organisms.
3. To demonstrate how computational and machine learning approaches can
   complement traditional microbiology laboratory data analysis.

### Data Source

The dataset used in this study consists of phenotypic antimicrobial
susceptibility records derived from publicly available antimicrobial
resistance databases. Each record includes information on the bacterial
species, antibiotic tested, and resistance phenotype (e.g., susceptible or
resistant).

This work is designed as an independent, hypothesis-driven mini research
project to demonstrate research readiness in microbiology and immunology,
integrating laboratory domain knowledge with computational analysis.

## Dataset Harmonization and Organism Annotation

To enable comparative analysis between bacterial species, an explicit
`Organism` label is added to each dataset prior to merging.
This allows downstream analyses to examine species-specific resistance
patterns while maintaining a unified data structure.

```python
# ADD ORGANISM LABELS
ecoli_df["Organism"] = "Escherichia coli"
kleb_df["Organism"] = "Klebsiella pneumoniae"
```

# MERGE DATASET

```python
amr_df = pd.concat([ecoli_df, kleb_df], ignore_index=True)
amr_df.shape
```

```python
amr_df["Organism"].value_counts()
```

## Distribution of Resistant Phenotypes Across Organisms

Understanding how resistance differs between *E. coli* and
*Klebsiella pneumoniae* is essential for both clinical treatment
and antimicrobial stewardship.

```python
amr_df["Resistant Phenotype"].value_counts()
```

```python
pd.crosstab(
    amr_df["Organism"],
    amr_df["Resistant Phenotype"],
    normalize="index"
)
```

K. pneumoniae shows a higher proportion of resistance compared to E. coli across tested antibiotics.

## Antibiotic-Specific Resistance Profiles

Antibiotic-level resistance patterns were examined to identify drugs
with consistently high resistance rates across organisms.

```python
top_antibiotics = amr_df["Antibiotic"].value_counts().head(10).index

subset = amr_df[amr_df["Antibiotic"].isin(top_antibiotics)]

pd.crosstab(
    subset["Antibiotic"],
    subset["Resistant Phenotype"],
    normalize="index"
).sort_values(by="Resistant", ascending=False)
```

## Distribution of Resistance Phenotypes by Organism

```python
plt.figure(figsize=(7,5))
sns.countplot(
    data=amr_df,
    x="Resistant Phenotype",
    hue="Organism",
    order=["Susceptible", "Intermediate", "Resistant"]
)

plt.title("Resistance Phenotype Distribution by Organism")
plt.xlabel("Resistance Phenotype")
plt.ylabel("Number of Isolates")
plt.legend(title="Organism")
plt.tight_layout()
plt.show()
```

## Key Findings

### Organism-Level Resistance Patterns

A substantial difference in antimicrobial resistance prevalence was observed
between the two bacterial species analyzed. *Klebsiella pneumoniae* exhibited
a markedly higher proportion of resistant isolates (≈62%) compared to
*Escherichia coli* (≈23%). This finding is consistent with the known propensity
of *K. pneumoniae* to acquire extended-spectrum β-lactamases (ESBLs) and
carbapenem resistance mechanisms.

### Antibiotic-Specific Resistance Trends

High resistance rates were observed for commonly prescribed antibiotics,
including ampicillin and third-generation cephalosporins such as ceftriaxone.
In contrast, aminoglycosides (e.g., amikacin) and carbapenems (e.g., meropenem)
demonstrated comparatively preserved susceptibility, highlighting their
continued clinical relevance in severe infections.

##  Proportional Resistance Profiles of E. coli and K. pneumoniae

```python
prop_df = pd.crosstab(
    amr_df["Organism"],
    amr_df["Resistant Phenotype"],
    normalize="index"
).reset_index()

prop_df = prop_df.melt(
    id_vars="Organism",
    var_name="Phenotype",
    value_name="Proportion"
)

plt.figure(figsize=(7,5))
sns.barplot(
    data=prop_df,
    x="Organism",
    y="Proportion",
    hue="Phenotype"
)

plt.title("Normalized Resistance Phenotypes by Organism")
plt.ylabel("Proportion of Isolates")
plt.xlabel("")
plt.legend(title="Phenotype")
plt.tight_layout()
plt.show()
```

## Resistance Rates of Commonly Tested Antibiotics

```python
top_ab = amr_df["Antibiotic"].value_counts().head(10).index

ab_df = amr_df[amr_df["Antibiotic"].isin(top_ab)]

resistance_rate = (
    pd.crosstab(ab_df["Antibiotic"], ab_df["Resistant Phenotype"], normalize="index")
    .reset_index()
)

plt.figure(figsize=(9,6))
sns.barplot(
    data=resistance_rate,
    x="Resistant",
    y="Antibiotic"
)

plt.title("Resistance Rates Across Commonly Tested Antibiotics")
plt.xlabel("Proportion Resistant")
plt.ylabel("Antibiotic")
plt.tight_layout()
plt.show()
```

## Heatmap of Resistance Rates by Organism and Antibiotic

```python
heatmap_df = pd.crosstab(
    amr_df["Antibiotic"],
    amr_df["Organism"],
    values=(amr_df["Resistant Phenotype"] == "Resistant"),
    aggfunc="mean"
)

heatmap_df = heatmap_df.loc[top_ab]

plt.figure(figsize=(6,7))
sns.heatmap(
    heatmap_df,
    annot=True,
    cmap="Reds",
    fmt=".2f"
)

plt.title("Resistance Proportion Heatmap")
plt.xlabel("Organism")
plt.ylabel("Antibiotic")
plt.tight_layout()
plt.show()
```
